```python
# no data yet for opposite reference stimulus

import numpy as np
import pandas as pd

scores = np.load('scores/S1/perceived_multispeaker/attend-M.npz', allow_pickle=True)
print(scores.files)
score_names = ['window_scores', 'window_zscores', 'story_scores', 'story_zscores']
score_files = {}
for name in score_names:
    score_files[name] = scores[name].item()

print(score_files['story_zscores'][('attend-M', 'BERT')])
```

```python
story_zscores = {'subject': [], 'task':[], 'BERT':[]}
for subject in [1,2,3]:
    for task in ['attend-M', 'attend-F']:
        scores = np.load(f'scores/S{subject}/perceived_multispeaker/{task}.npz', allow_pickle=True)['story_zscores'].item()
        story_zscores['subject'].append(subject)
        story_zscores['task'].append(task)
        story_zscores['BERT'].append(scores[(task, 'BERT')])

for subject in [1,2,3]:
    for task in ['attend-M', 'attend-F']:
        scores = np.load(f'scores_diff_ref_pm/S{subject}/perceived_multispeaker/{task}.npz', allow_pickle=True)['story_zscores'].item()
        story_zscores['subject'].append(subject)
        story_zscores['task'].append(task + '-ref-F') if task[-1] == 'M' else story_zscores['task'].append(task + '-ref-M')
        story_zscores['BERT'].append(scores[('attend-F', 'BERT')]) if task[-1] == 'M' else story_zscores['BERT'].append(scores[('attend-M', 'BERT')])

story_zscores
```

```python
story_zscores_df = pd.DataFrame(story_zscores)
story_zscores_df.to_csv('perceived_multispeaker_zscores.csv', index=False)

story_zscores_df
```

```python
window_zscores = {'subject': [], 'task':[], 'BERT':[]}
for subject in [1,2,3]:
    for task in ['attend-M', 'attend-F']:
        scores = np.load(f'scores/S{subject}/perceived_multispeaker/{task}.npz', allow_pickle=True)['window_zscores'].item()
        window_zscores['subject'].append(subject)
        window_zscores['task'].append(task)
        window_zscores['BERT'].append(scores[(task, 'BERT')])

for subject in [1,2,3]:
    for task in ['attend-M', 'attend-F']:
        scores = np.load(f'scores_diff_ref_pm/S{subject}/perceived_multispeaker/{task}.npz', allow_pickle=True)['window_zscores'].item()
        window_zscores['subject'].append(subject)
        window_zscores['task'].append(task + '-ref-F') if task[-1] == 'M' else window_zscores['task'].append(task + '-ref-M')
        window_zscores['BERT'].append(scores[('attend-F', 'BERT')]) if task[-1] == 'M' else window_zscores['BERT'].append(scores[('attend-M', 'BERT')])

window_zscores
```

```python
results_df = pd.DataFrame(window_zscores)

print(len(results_df.loc[0, 'BERT']))
print(len(results_df.loc[1, 'BERT']))
print('=')
print(315*2)
m=630

from scipy.stats import norm
#convert to p val
results_df['BERT'] = results_df['BERT'].apply(lambda l : [1-norm.cdf(z) for z in l])
#sort
for i in range(6):
    results_df.loc[i,'BERT'].sort()
# to q and check threshold
def p_to_q(p, i):
    return p*m/i
results_df['BERT'] = results_df['BERT'].apply(lambda l : [1 if p_to_q(l[i-1], i) < 0.05 else 0 for i in range(1, len(l)+1)])

results_df['percent_sig_decoded'] = results_df['BERT'].apply(lambda l : np.array(l).mean())
results_df.drop(columns=['BERT'], inplace=True)
# results_df['BERT_zscore'] = story_zscores['BERT']

print('YET TO DO T-TEST')
results_df
```

```python
results_df.to_csv('perceived_multispeaker_percentages.csv', index=False)
results_df
```
