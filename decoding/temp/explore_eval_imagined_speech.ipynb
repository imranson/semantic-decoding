```python
# imagined speech, fig 3a, 1 prediction similarity to stimulus vs 1 prediction similaairity to incorrect stimulus, pairwise accuracy, can only classify, not significant

# no data yet for incorrect transcript's similarity

import numpy as np
import pandas as pd

scores = np.load('scores_bravo_ref_ims/S1/imagined_speech/bravo_repeat-1.npz', allow_pickle=True)
print(scores.files)
score_names = ['window_scores', 'window_zscores', 'story_scores', 'story_zscores']
score_files = {}
for name in score_names:
    score_files[name] = scores[name].item()

print(score_files['story_zscores'])
print(score_files['story_zscores'][('bravo', 'BERT')])
```

```python
zscores_df = pd.DataFrame({'task':[], 'alpha_ref':[],'bravo_ref':[],'charlie_ref':[], 'delta_ref':[], 'echo_ref':[]})

to_add = {'task':[], 'alpha_ref':[],'bravo_ref':[],'charlie_ref':[], 'delta_ref':[], 'echo_ref':[]}
for task in ['alpha','bravo', 'charlie', 'delta', 'echo']:
    for repeat in [1,2]:
        to_add['task'].append(f'{task}_repeat-{repeat}')
        for ref in ['alpha','bravo', 'charlie', 'delta', 'echo']:
            scores = np.load(f'scores_{ref}_ref_ims/S1/imagined_speech/{task}_repeat-{repeat}.npz', allow_pickle=True)
            # print(scores['story_zscores'].item()[(ref, 'BLEU')])
            to_add[f'{ref}_ref'].append(scores['story_zscores'].item()[(ref, 'BERT')])

zscores_df = pd.DataFrame(to_add)

zscores_df
```

```python
zscores_df.to_csv('imagined_speech_zscores.csv', index=False)

zscores_df
```
